## Question 2.3 - How did the number of addition changed in earlier and later years.

```python
yearly_counts = df1['added_year'].value_counts().sort_index()
yearly_counts
```

## Totals

```python
later_total = yearly_counts.loc[2015:2020].sum()
later_total
```

```python
earlier_total = yearly_counts.loc[2008:2014].sum()
earlier_total
```

### Calculating Average

```python
len(yearly_counts.loc[2008:2014])
```

```python
eavg = earlier_total/len(yearly_counts.loc[2008:2014])
eavg
```

```python
len(yearly_counts.loc[2015:2020])
```

```python
lavg = later_total/len(yearly_counts.loc[2015:2020])
lavg
```

## Calculating Percentage of Increase

```python
title_addition_percentage = round((lavg - eavg) / eavg * 100, 2)
title_addition_percentage
```

## Question 2.4 - Which periods saw the largest increases or decreases?

```python
difference_count = yearly_counts.diff()
difference_count
```

```python
max_difference_val = difference_count.max()
max_difference_index = difference_count.idxmax()
```

```python
min_difference_val = difference_count.min()
min_difference_index = difference_count.idxmin()
```

```python
print(f'Most titles were added in: {max_difference_index} -> {max_difference_val}')
```

```python
print(f'Least titles were added in: {min_difference_index} -> {min_difference_val}')
```
